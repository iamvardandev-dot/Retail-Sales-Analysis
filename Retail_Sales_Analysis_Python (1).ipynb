# Profit

```python
df['Profit']=df['Sales']-df['Cost']
```

```python
df.head(2)
```

# Which product generates the highest total profit?

```python
df.groupby(['Product'])['Profit'].sum()
```

```python
df['Profit_Margin']=(df['Profit']/df['Sales'])*100
```

```python
df.head(2)
```

# What is the average profit margin for each product?

```python
df.groupby(['Product'])['Profit_Margin'].mean()
```

# What is the average profit margin for each category?

```python
df.groupby(['Category'])['Profit_Margin'].mean()
```
